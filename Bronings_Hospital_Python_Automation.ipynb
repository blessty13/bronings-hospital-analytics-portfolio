```python
import pandas as pd
import numpy as np
```

```python
#load outpatient data files
appointments =pd.read_csv("/03_outpatient_appointments.csv")

print(appointments)

#Remove dupliacate records.
appointments= appointments.drop_duplicates()

#Clean column names
appointment.column= appointments.columns.str.strip()

# Replace blank values with NaN
appointment.replace(r'^\s*$', np.nan, regex=True, inplace=True)

# Convert dates
appointments["Appointment_Date"] = pd.to_datetime(
    appointments["Appointment_Date"],
    errors="coerce")

# Convert waiting days to numeric
appointments["Waiting_Days"] = pd.to_numeric(
    appointments["Waiting_Days"],
    errors="coerce")

# Convert RTT weeks to numeric
appointments["RTT_Weeks"] = pd.to_numeric(
    appointments["RTT_Weeks"],
    errors="coerce")
```

```python
#DNA Rate By Speciality Calculation
dna_summary = (appointments
    .groupby("Specialty_Code")
    .apply(lambda x: pd.Series({"Total_Appointments": len(x),"DNA_Count":
            (x["Attendance_Status"].str.upper() == "DNA").sum(),"DNA_Rate (%)":
            round(((x["Attendance_Status"].str.upper() == "DNA").sum()/ len(x)
            ) * 100,
                2)}))
    .reset_index())

print(dna_summary)
```

```python
#RTT Breach Count code
rtt_breach_count =(appointments["RTT_Weeks"] > 18).sum()

print("RTT Breach Count:", rtt_breach_count)
```

```python
#Data Quality (DQ) Score function
def dq_score(df):
    total_cells = df.shape[0] * df.shape[1]
    missing_cells = df.isnull().sum().sum()

    score = ((total_cells - missing_cells)
        / total_cells) * 100

    return round(score, 2)

dq = dq_score(appointments)

print("Data Quality Score:", dq)
```

```python
#Excel Output
with pd.ExcelWriter("Hospital_Analytics_Report.xlsx") as writer:

    appointments.to_excel(
        writer,
        sheet_name="Cleaned_Data",
        index=False
    )

    dna_summary.to_excel(
        writer,
        sheet_name="DNA_By_Specialty",
        index=False
    )

print("Excel report created successfully.")
```
